```python
print("🏥 HealthSense AI - BMI Calculator")
print("----------------------------------")

name = input("Enter Your Name: ")
age = int(input("Enter Your Age: "))
weight = float(input("Enter Weight (kg): "))
height = float(input("Enter Height (cm): "))

# Convert cm to meters
height_m = height / 100

# Calculate BMI
bmi = weight / (height_m * height_m)

print("\n📊 RESULTS")
print("----------------------------------")
print("Name:", name)
print("Age:", age)
print("BMI:", round(bmi, 2))

# BMI Category & Health Tips
if bmi < 18.5:
    print("Status: Underweight")
    print("💡 Health Tip:")
    print("Eat nutritious foods and increase calorie intake.")

elif bmi < 25:
    print("Status: Normal Weight")
    print("💡 Health Tip:")
    print("Great job! Maintain your healthy lifestyle.")

elif bmi < 30:
    print("Status: Overweight")
    print("💡 Health Tip:")
    print("Exercise regularly and follow a balanced diet.")

else:
    print("Status: Obese")
    print("💡 Health Tip:")
    print("Consider consulting a healthcare professional.")

print("----------------------------------")
print("✅ Thank you for using HealthSense AI")

```
